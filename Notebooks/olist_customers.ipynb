```python
df["customer_state"].value_counts().head(10).plot(kind="line", title="Customer State Distribution")
plt.xlabel("Customer State")
plt.ylabel("Count")
plt.show()
```

```python
df["customer_city"].value_counts().head(10).plot(kind="bar", title="Customer State Distribution")
plt.xlabel("Customer city")
plt.ylabel("Count")
plt.show()
```
