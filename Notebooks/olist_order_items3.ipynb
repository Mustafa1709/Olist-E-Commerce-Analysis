```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("olist_order_items_dataset_cle.csv")
```

```python
df.head(10)
```

```python
df.isnull().sum()
```

```python
df=df.rename(columns={" price ":"price"})
```

```python
df=df.rename(columns={" freight_value ":"freight_value"})
```

```python
df.dtypes
```

```python
df["freight_value"] = pd.to_numeric(
    df["freight_value"],
    errors="coerce"
)
```

```python
date_columns = [
    "shipping_limit_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        format = "%d-%m-%Y %H:%M",
        errors="coerce"
    )
```

```python
df["shipping_year"]=df["shipping_limit_date"].dt.year 
```

```python
df["shipping_month"]=df["shipping_limit_date"].dt.month_name()
```

```python
df["total_value"] = df["price"] + df["freight_value"].fillna(0)
```

```python
df.to_csv("olist_orders_items_cleaned2.csv", index=False)
```
