```python
import pandas as pd
```

```python
orders = orders = pd.read_csv("olist_orders_cleaned_dataset.csv")
customers = pd.read_csv("olist_customers_dataset_cle.csv")
sellers = pd.read_csv("olist_sellers_dataset_cle.csv",encoding="cp1252")
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql://postgres:Password@localhost:5432/Olist_db"
)
```

```python
orders.to_sql(
    name='orders',
    con=engine,
    if_exists='replace',
    index=False
)
```

```python
customers.to_sql(
    name='customers',
    con=engine,
    if_exists='replace',
    index=False
)
```

```python
sellers.to_sql(
    name='sellers',
    con=engine,
    if_exists='replace',
    index=False
)
```

```python
products = pd.read_csv("olist_products_dataset_cle.csv")
order_items = pd.read_csv("olist_orders_items_cleaned.csv")
```

```python
products.to_sql(
    name='products',
    con=engine,
    if_exists='replace',
    index=False
)
```

```python
order_items.to_sql(
    name='order_items',
    con=engine,
    if_exists='replace',
    index=False
)
```

```python
reviews = pd.read_csv("olist_order_reviews_dataset_cle.csv")
payments = pd.read_csv("olist_order_payments_dataset_cle.csv")


reviews.to_sql(
    name='reviews',
    con=engine,
    if_exists='replace',
    index=False
)

payments.to_sql(
    name='payments',
    con=engine,
    if_exists='replace',
    index=False
)
```
