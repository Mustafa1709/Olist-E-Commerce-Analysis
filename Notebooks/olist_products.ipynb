product category name wise average weight,length,heigh and width

```python
df.groupby("product_category_name")[[
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]].mean()
```

```python
df.columns = df.columns.str.lower()
```

product category name wise mean,median,min and max weight,length,heigh and width

```python
df.groupby("product_category_name")[[
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]].agg(["mean", "median", "min", "max"])
```

```python
df.groupby("product_category_name")[[
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]].mean().sort_values(
    by="product_weight_g",
    ascending=False
).round(2)
```

```python
df.groupby("product_category_name")["product_id"].count().sort_values(ascending=False)
```

product category wise count of products,avg weight and length 

```python
df.groupby("product_category_name").agg(
    product_count=("product_id", "count"),
    avg_weight=("product_weight_g", "mean"),
    avg_length=("product_length_cm", "mean")
).sort_values(by="product_count", ascending=False)
```

```python
df.groupby("product_category_name").agg(
    maximum_weight = ("product_weight_g","max")
).sort_values (by = "maximum_weight",ascending = False)
```

```python
df.groupby("product_category_name").agg(
    max_weight = ("product_weight_g","max"),
    min_weight = ("product_weight_g","min"),
    mean_weight = ("product_weight_g","mean"),
    product_count = ("product_id","count")
).sort_values(by = "product_count",ascending = False)
```

```python
df["product_volume_cm3"] = (
    df["product_length_cm"]*
    df["product_height_cm"]*
    df["product_width_cm"]
)
```

```python
df.groupby("product_category_name").agg(
    average_volume = ("product_volume_cm3","mean")
).sort_values("average_volume",ascending = False).round(2)
```

```python
df["product_volume_cm3"] = (
    df["product_length_cm"] *
    df["product_height_cm"] *
    df["product_width_cm"]
)

df.groupby("product_category_name").agg(
    avg_length=("product_length_cm", "mean"),
    avg_height=("product_height_cm", "mean"),
    avg_width=("product_width_cm", "mean"),
    avg_weight=("product_weight_g", "mean"),
    avg_volume=("product_volume_cm3", "mean")
).sort_values(by="avg_volume", ascending=False).round(2)
```

```python
for col in [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(col, len(outliers))
```
