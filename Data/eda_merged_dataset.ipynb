```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("olist_master_dataset.csv")
```

```python
df.head(10)
```

```python
df.columns
```

```python
# Did sales Grow overtime ?

df.groupby("purchased_year")["payment_value"].sum().plot(
    kind = "bar",
    color = "teal",
    title="Distribution",
    figsize=(5,4)
)
```

Observation
Yes sales grown over years starting from 2016 to 2018

```python
df[["payment_value", "price", "freight_value"]].dtypes
```

```python
df["payment_value"] = pd.to_numeric(df["payment_value"], errors="coerce")
```

```python
bi = df.groupby("purchased_month")["payment_value"].sum().sort_values(ascending=False).plot(
    kind = "bar",
    figsize=(5,4),
    fontsize=8,
    color = "Teal"

)
for container in bi.containers:
    bi.bar_label(container, fmt="%.1f%%",fontsize=6)
bi.spines["right"].set_visible(False)
bi.spines["top"].set_visible(False)  
bi.spines["left"].set_visible(False) 

bi.set_xlabel("Months")
bi.set_title("Month wise Revenue",fontsize = 10,fontfamily = "Arial")

bi.set_yticks([])
```
